```python
display(df.head(10))

print('\nDescriptive statistics for the student DataFrame:')
display(df.describe())

print('\nReshaping scores using StudentID as index:')
df_indexed = df.set_index('StudentID')

df_stacked_scores = df_indexed[['ScoreSE', 'ScoreML']].stack()
print('Stacked scores (Series with MultiIndex):')
display(df_stacked_scores.head())

df_unstacked_scores = df_stacked_scores.unstack()
print('\nUnstacked scores:')
display(df_unstacked_scores.head())

print('\nConcatenating and unstacking score series:')
series_score_se = df_indexed['ScoreSE'].dropna()
series_score_ml = df_indexed['ScoreML'].dropna()

combined_scores_series = pd.concat([series_score_se, series_score_ml], keys=['SE_Scores', 'ML_Scores'])
print('Combined scores (Series with MultiIndex from concatenation):')
display(combined_scores_series.head())

combined_scores_df = combined_scores_series.unstack(level=0)
print('\nUnstacked combined scores:')
display(combined_scores_df.head())
```
