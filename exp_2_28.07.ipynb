```python

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer
```

```python

employees = pd.DataFrame({
    "Age":        [28, np.nan, 35, 41, 24],
    "Salary":     [42000, 55000, np.nan, 71000, 39000],
    "Department": ["Sales", "IT", "Sales", np.nan, "IT"],
})
employees
```

```python
mean_imputer = SimpleImputer(strategy="mean")

employees["Age_Mean"] = mean_imputer.fit_transform(employees[["Age"]]).ravel()
employees[["Age", "Age_Mean"]]
```

```python

median_imputer = SimpleImputer(strategy="median")
employees["Salary_Median"] = median_imputer.fit_transform(employees[["Salary"]]).ravel()
employees[["Salary", "Salary_Median"]]
```

```python

mode_imputer = SimpleImputer(strategy="most_frequent")
employees["Department_Mode"] = mode_imputer.fit_transform(employees[["Department"]]).ravel()
employees[["Department", "Department_Mode"]]
```

```python
num_cols = ["Age", "Salary"]

mice_imputer = IterativeImputer(max_iter=10, random_state=42)

mice_result = employees[num_cols].copy()
mice_result.iloc[:, :] = mice_imputer.fit_transform(mice_result)

employees["Age_MICE"] = mice_result["Age"]
employees["Salary_MICE"] = mice_result["Salary"]
```
